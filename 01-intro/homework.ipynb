## Load the data

Read the course's pinned 2026 release directly from its authoritative source.

```python
df = pd.read_csv(DATA_URL)
df.head()
```

## Q2. Records count

```python
record_count = len(df)
record_count
```

**Answer: 10000**

## Q3. Fuel types

```python
fuel_type_count = df['fuel_type'].nunique()
fuel_type_count
```

**Answer: 3**

## Q4. Missing values

```python
missing_by_column = df.isna().sum()
missing_by_column[missing_by_column > 0]
```

```python
columns_with_missing_values = df.isna().any().sum()
columns_with_missing_values
```

**Answer: 2**

## Q5. Maximum fuel efficiency for Asian cars

```python
asia_max_fuel_efficiency = df.loc[
    df['origin'].eq('Asia'), 'fuel_efficiency_mpg'
].max()
asia_max_fuel_efficiency
```

**Answer: 41.2**

## Q6. Median horsepower after imputing the mode

```python
horsepower_median_before = df['horsepower'].median()
horsepower_mode = df['horsepower'].mode().iloc[0]
horsepower_median_after = df['horsepower'].fillna(horsepower_mode).median()

{
    'median_before': horsepower_median_before,
    'mode_used_for_imputation': horsepower_mode,
    'median_after': horsepower_median_after,
}
```

The median falls from 254 to 252 after filling missing values with the mode.

**Answer: Yes, it decreased.**

## Q7. Sum of weights

```python
X = (
    df.loc[df['origin'].eq('Asia'), ['vehicle_weight', 'model_year']]
    .head(7)
    .to_numpy()
)
XTX = X.T @ X
XTX_inverse = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inverse @ X.T @ y

{'weights': w, 'sum_of_weights': w.sum()}
```

**Answer: 0.369**
